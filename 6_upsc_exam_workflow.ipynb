```python
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, START, END
from dotenv import load_dotenv
from typing import TypedDict, Annotated
from pydantic import BaseModel, Field
import operator
```

```python
load_dotenv()
```

```python
model = ChatOpenAI(model='gpt-4o-mini')
```

```python
class EvaluationSchema(BaseModel):

    feedback: str = Field(description="Detailed feedback for the essay")
    score: int = Field(description="Score out of 10", ge=0, le=10)
```

```python
structured_output = model.with_structured_output(EvaluationSchema)
```

```python
essay = """"""
```

```python
prompt = f'Evaluate the following essay with respect to language, clarity of thought and depth of analysis. Provide detailed feedback and a score out of 10. {essay}'
structured_output.invoke(prompt)
```

```python
# To extract specific field, use this

structured_output.invoke(prompt).feedback
structured_output.invoke(prompt).score
```
